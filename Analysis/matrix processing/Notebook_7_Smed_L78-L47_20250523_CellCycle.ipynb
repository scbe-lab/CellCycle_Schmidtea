## Calculating neoblast enrichment for each gene (marker function)

```python
sc.tl.rank_genes_groups(adata, groupby= 'broad_names', method="wilcoxon", use_raw=True, key_added = 'broad')
```

```python
neo_enrich_df = pd.concat([
    pd.DataFrame(adata.uns['broad']['names'])['neoblasts'].rename('names'),
    pd.DataFrame(adata.uns['broad']['scores'])['neoblasts'].rename('scores'),
    pd.DataFrame(adata.uns['broad']['pvals'])['neoblasts'].rename('pvals'),
    pd.DataFrame(adata.uns['broad']['pvals_adj'])['neoblasts'].rename('pvals_adj'),
    pd.DataFrame(adata.uns['broad']['logfoldchanges'])['neoblasts'].rename('logfoldchanges'),
    
], axis = 1).set_index('names')
```

```python
neo_enrich_df
```

```python
neoblast_list = neo_enrich_df[(neo_enrich_df['logfoldchanges'] > 0) & (neo_enrich_df['pvals_adj'] < 0.05)].index.to_list()
```

```python
len(neoblast_list)
```

```python
sc.pl.umap(adata, color = sample(neoblast_list, 40), cmap = umap_cmap, size = 10)
```

# Intersect with the list of differentially expressed genes between G1 and G2 samples

```python
# this file is the results from the DGE analysis between G1 and G2 neoblasts (see R markdown: 01_G1_vs_G2_DGE.Rmd)
df_cc = pd.read_excel('DGE/2C_4C/outputs/G1_vs_G2_diffExpGenes.xlsx', index_col='geneID')
```

```python
G1_li = df_cc[df_cc['overexpressed_in'] =='G1'].index.to_list()
```

```python
G2_li = df_cc[df_cc['overexpressed_in'] =='G2'].index.to_list()
```

```python
# intersect between the results of the DGE and genes expressed in neoblasts
G1_li_f = [i for i in G1_li if i in neoblast_list]
G2_li_f = [i for i in G2_li if i in neoblast_list]
```

```python
adata.uns.keys()
```

```python
adata.uns['2c_li_DGE'] = G1_li_f 
adata.uns['4c_li_DGE'] = G2_li_f 
```

```python
sc.pl.umap(adata, color = G1_li_f, cmap = umap_cmap, size = 10)
```

```python
sc.pl.umap(adata, color = G2_li_f, cmap = umap_cmap, size = 10)
```

```python
sc.tl.score_genes(adata, G1_li_f, score_name='score_DGE_G1')
sc.tl.score_genes(adata, G2_li_f, score_name='score_DGE_G2')
```

```python
# Violin plot for the broad groups with conditions side by side
# Subset to the conditions of interest
subset = adata[adata.obs['Condition'].isin(['GFP','Cdh1','H2B'])].copy()
subset.obs['Condition_broad'] = (
    subset.obs['broad_names2'].astype(str) + '_' + subset.obs['Condition'].astype(str)
)

# Define the color palette
cond_colors = {
    'GFP' : 'slategray',  
    'Cdh1': 'dodgerblue',   
    'H2B' : 'darkviolet'   
}
subset.obs['Condition_broad'] = subset.obs['Condition_broad'].astype('category')
cat_order = subset.obs['Condition_broad'].cat.categories
palette = [cond_colors[cat.split('_')[-1]] for cat in cat_order]

# Plot violin
with plt.rc_context({'figure.figsize': (15, 5)}):
    sc.pl.violin(
        subset,
        'score_DGE_G2',
        groupby='Condition_broad',
        jitter=False,
        rotation=90,
        stripplot=False,
        palette=palette
    )
```
