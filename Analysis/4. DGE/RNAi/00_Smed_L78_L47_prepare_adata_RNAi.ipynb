```python
# new adata object with the unprocessed counts
```

```python
filt = adata_p.obs['Experiment'] == 'RNAi'
filt.values
```

```python
X_ = adata.X[filt.values, :]
adata_rnai = ad.AnnData(X_)
```

```python
# .obs and .var from the processed object
adata_rnai.obs = adata_p.obs[filt]
adata_rnai.obs.index = adata_p.obs.index[filt]
adata_rnai.var = adata_p.raw.var
```

```python
adata_rnai
```

```python
# Extract the obs and vars
adata_rnai.obs.to_csv('./rnai_cells_metadata_for_DGE.csv') # obs
adata_rnai.var.to_csv('./rnai_all_genes_metadata_for_DGE.csv') # var, all
```

```python
# Get the raw counts matrix
raw_counts = adata_rnai.X
```

```python
# Define the output file path
output_file = "./rnai_raw_for_DGE.mtx.gz"

# Save the raw counts matrix as .mtx.gz
with gzip.open(output_file, 'wb') as f:
    mmwrite(f, raw_counts)
```
