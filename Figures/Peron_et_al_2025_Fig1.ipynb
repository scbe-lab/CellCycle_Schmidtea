## neoblast score

```python
with plt.rc_context({'figure.figsize': (10, 10)}):
    sc.pl.umap(adata, color='neoblast_score', frameon=False, size= 20, title = '', color_map = 'Purples', save= figure + '_neoblast-score.pdf')
```

```python
# Create and sort DataFrame
score = pd.DataFrame(list(adata.uns['neoblast_score_annotated_names'].items()), columns=['Cell Type', 'neoblast_score'])
data = score.sort_values('neoblast_score', ascending=False).reset_index(drop=True)

# color map
color_map = dict(zip(adata.obs['annotated_names'].cat.categories, adata.uns['annotated_names_colors']))
li_colours = [color_map[name] for name in data['Cell Type']]
    
# Plot
with plt.rc_context({'figure.figsize': (15, 7)}):
    sns.barplot( x='Cell Type', y='neoblast_score', data=data, palette=li_colours )
    plt.xticks(rotation=90)
    plt.axvline(9.5, 0, color = 'black')
    plt.axvline(34.5, 0, color = 'black')
    plt.title('Neoblast Score per Cluster', fontsize=16, fontweight='bold')
    plt.xlabel('')
    plt.ylabel('Neoblast Score')
    plt.tight_layout()
    plt.savefig('./Figure_plots/'+ figure + '/' + figure + "_barplot_neoblastscore.pdf")
    plt.show()


```

```python
clusteringbroad = 'broad_names'
```

```python
broad = clusteringbroad
narrow = clusteringlayer

line_w = 1


dfb = pd.DataFrame(index = adata.obs[broad].cat.categories, data = adata.obs[broad].value_counts(normalize = True) * 100)
dfn = pd.DataFrame(index = adata.obs[narrow].cat.categories, data = adata.obs[narrow].value_counts(normalize = True) * 100)

colours_b = list(adata.uns[dfb.columns[0] + '_colors'])
colours_n = list(adata.uns[dfn.columns[0] + '_colors'])

fig, axs = plt.subplots(1, 2, figsize=(10, 15))


sns.barplot(data = dfb, y = dfb.index, x = dfb.columns[0], hue = dfb.index, palette = colours_b, ax = axs[0])

axs[0].set_ylabel(None)
axs[0].spines["left"].set_linewidth(line_w)
axs[0].spines["bottom"].set_linewidth(line_w)
axs[0].spines["top"].set_linewidth(line_w)
axs[0].spines["right"].set_linewidth(line_w)

axs[0].tick_params(axis='both',       # apply to x and y axes
               which='major',     # major ticks
               width=line_w,           # thickness
               length=4)          # length in points

for ticklabel, color in zip(axs[0].get_yticklabels(), colours_b):
    ticklabel.set_color(color)

for i in range(1, len(dfb.index)):
    axs[0].axhline(y= i - 0.5, color = 'black', linewidth = line_w)

sns.barplot(data = dfn, y = dfn.index, x = dfn.columns[0], hue = dfn.index, palette = colours_n, ax = axs[1])
axs[1].set_ylabel(None)
axs[1].spines["left"].set_linewidth(line_w)
axs[1].spines["bottom"].set_linewidth(line_w)
axs[1].spines["top"].set_linewidth(line_w)
axs[1].spines["right"].set_linewidth(line_w)



axs[1].tick_params(axis='both',       # apply to x and y axes
               which='major',     # major ticks
               width=line_w,           # thickness
               length=4)          # length in points

for ticklabel, color in zip(axs[1].get_yticklabels(), colours_n):
    ticklabel.set_color(color)

value = 0
for i in adata.obs[clusteringbroad].cat.categories:
    value = value + len(adata.obs.loc[adata.obs[clusteringbroad] == i, clusteringlayer].unique())
    axs[1].axhline(y= value - 0.5, color = 'black', linewidth = line_w)


plt.tight_layout()
plt.savefig('./Figure_plots/'+ figure + '/' + figure + "_barplot_broad_narrow.pdf")
plt.show()
```
