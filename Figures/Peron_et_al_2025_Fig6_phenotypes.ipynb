## phenotypes

the numbers come from observation of the intact and regeneration phenotypes

```python
cmap1 = ['grey', 'lightgray', 'rebeccapurple', 'mediumpurple', 'royalblue', 'lightsteelblue']
```

```python
cmap2 = ['#708090', '#a9a9a9', '#9400d3', '#ba55d3', '#1e90ff', '#87ceeb']
```

```python
#GFP, H2B, Cdh1
# separated replicates
dico = { 'no phenotype': [46, 54, 21, 43, 26, 37], 
        'head regression': [0, 0, 11, 4, 0, 0],
        'dark lesions': [0, 0, 13, 2, 12, 12],
        'dead': [0, 0, 5, 1, 12, 1]}
```

```python
df = pd.DataFrame(dico).T
df.rename(columns={0: 'gfp 1', 1: 'gfp 2', 2:'h2b 1', 
                   3: 'h2b 2', 4: 'cdh 1', 5:'cdh 2'}, inplace=True)
```

```python
df = (df / df.sum(axis=0) )*100
df
```

```python
fig, ax = plt.subplots(figsize=(7, 6))
df.T.plot(kind='bar', stacked=True, width=0.8, color=['grey', 'lightgray', 'white', 'black'], edgecolor = ['black'], ax=ax)

# ax.set_ylabel("Percentage of worms", fontsize=20)
ax.tick_params(axis='x', rotation=45, labelsize=20)
ax.tick_params(axis='y', labelsize=20)

# Reverse legend order (bottom → top)
handles, labels = ax.get_legend_handles_labels()
ax.legend(
    handles[::-1],
    labels[::-1],
    bbox_to_anchor=(1, 1),
    loc="upper left",
    prop={'size': 20}
)


plt.savefig("Figure_plots/" + figure + '/ACME_phenotypes.pdf', bbox_inches="tight")
plt.show()
```

```python

plt.rcParams['font.family'] = 'Arial'

df = pd.DataFrame(dico_cyto)
grp = df.groupby(['treatment','replicate'], sort=False)['percentage G2'].agg(['mean','std']).reset_index()

colors = ['grey', 'lightgray', 'rebeccapurple', 'mediumpurple', 'royalblue', 'lightsteelblue']
x = np.array([0, 0.5, 1.5, 2.0, 3.0, 3.5])

plt.figure(figsize=(5, 6))
plt.bar(x, grp['mean'], width=0.45, color=colors, edgecolor='black',
        yerr=grp['std'], capsize=4)
plt.ylabel('Percentage G2', fontsize=20)
plt.xticks([])
plt.tick_params(axis='x', length=0)
plt.yticks(fontsize=18)
plt.savefig("Figure_plots/" + figure + '/cytometry.pdf', bbox_inches="tight")
plt.show()
```
