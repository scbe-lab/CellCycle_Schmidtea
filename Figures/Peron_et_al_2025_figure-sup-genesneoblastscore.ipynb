```python
import numpy as np
import pandas as pd
import scanpy as sc
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sc.settings.verbosity = 3 
sc.logging.print_header()
sc.settings.set_figure_params(dpi=300, transparent = True, format = 'pdf', vector_friendly = True)
```

```python
figure = "Figure_sup_genesneoblastscore/"
```

```python
sc.settings.figdir = './Figure_plots/'+figure
```

```python
umap_cmap = sns.blend_palette(['xkcd:light grey', 'xkcd:blueberry'], as_cmap = True)
```

# Declaring the input files

```python
adata = sc.read_h5ad('./h5ad/analysis_250528_f/Smed_L78-L47_20250523_Annotated.h5ad')
```

```python
adata
```

```python
samp = 'Sample'
```

```python
clusteringlayer = 'annotated_names'
```

## genes neoblast score

```python
markers_w = pd.DataFrame(adata.uns['rank_genes_groups_wilcox_leiden_2.5']['names']).head(50)
li_neo = list(set(markers_w['0'].head(50).to_list() + markers_w['1'].head(50).to_list()))
```

```python
len(li_neo)
```

```python
import matplotlib.pyplot as plt

plt.rcParams['figure.dpi'] = 300

fig = sc.pl.umap(
    adata,
    color=li_neo,
    cmap=umap_cmap,
    ncols=8,
    size=8,
    frameon=True,
    return_fig=True
)

for ax in fig.axes:
    ax.set_xlabel('')
    ax.set_ylabel('')
    ax.set_title(ax.get_title(), fontsize=20, fontweight='bold')

fig.savefig(
    f'./Figure_plots/{figure}/{figure}_genes-neo_umap.pdf',
    bbox_inches='tight'
)

plt.show()
```

```python
import matplotlib.pyplot as plt

# Get the category order from adata
category_order = adata.obs[clusteringlayer].cat.categories.tolist()

dp = sc.pl.dotplot(
    adata,
    li_neo,
    groupby=clusteringlayer,
    swap_axes=True,
    cmap=umap_cmap,
    vmax=1,
    vmin=0.05,
    dot_min=0.1,
    dot_max=1,
    standard_scale='var',
    figsize=(22, 10),
    categories_order=category_order,  # enforces x-axis order
    return_fig=True,
)

dp.largest_dot = 80
dp.make_figure()

ax = dp.ax_dict['mainplot_ax']
ax.tick_params(axis='x', labelsize=12)
ax.tick_params(axis='y', labelsize=12)

plt.savefig( './Figure_plots/'+ figure+ 'genes-neo_dotplot.pdf')
plt.show()
```
