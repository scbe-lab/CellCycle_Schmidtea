```python
import numpy as np
import pandas as pd
import scanpy as sc
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sc.settings.verbosity = 3 
sc.logging.print_header()
sc.settings.set_figure_params(dpi=300, transparent = True, format = 'pdf', vector_friendly = True)
```

```python
figure = "Figure_sup_umapviolin"
```

```python
sc.settings.figdir = './Figure_plots/'+figure
```

```python
umap_cmap = sns.blend_palette(['xkcd:light grey', 'xkcd:blueberry'], as_cmap = True)
```

# Declaring the input files

```python
adata = sc.read_h5ad('./h5ad/analysis_250528_f/Smed_L78-L47_20250523_Annotated.h5ad')
```

```python
adata
```

```python
samp = 'Sample'
```

```python
clusteringlayer = 'annotated_names'
```

# Plots

## umaps

```python
with plt.rc_context({'figure.figsize': (10, 10)}):
    sc.pl.umap(adata, color='n_genes', legend_loc='on data', legend_fontoutline = 5, title= 'n genes', size = 35,
        frameon=False, add_outline = False, cmap = 'Purples', save= figure + '_n-genes.pdf' )
```

```python
with plt.rc_context({'figure.figsize': (10, 10)}):
    sc.pl.umap(adata, color='total_counts', legend_loc='on data', legend_fontoutline = 5, title= 'total counts', size = 35,
        frameon=False, cmap = 'Purples', save= figure + '_tot-counts.pdf')
```

## violin plots

```python
with plt.rc_context({'figure.figsize': (20, 5)}):
    sc.pl.violin(adata, keys = "n_genes" , groupby = clusteringlayer, jitter = False, rotation = 90)
```

```python
with plt.rc_context({'figure.figsize': (20, 5)}):
    sc.pl.violin(adata, keys = "total_counts" , groupby = clusteringlayer, jitter = False, rotation = 90)
```

```python
fig, axs = plt.subplots(2, 1, figsize=(20, 10), sharex=True)  # sharex: same x-axis

# 1st violin plot
sc.pl.violin( adata, keys='n_genes', groupby=clusteringlayer, stripplot=False, rotation=90, ax=axs[0], show=False)
axs[0].set_xlabel('')  # remove x-axis labels
axs[0].set_ylabel('Number of genes', fontsize=14, fontweight='bold')

# 2nd violin plot
sc.pl.violin( adata, keys='total_counts', groupby=clusteringlayer, stripplot=False, rotation=90, ax=axs[1], show=False )
axs[1].set_ylabel('Total counts', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.savefig('./Figure_plots/'+figure + '/'+ figure + '_violin_plots.pdf', bbox_inches='tight')
plt.show()
```

```python
sc.pl.violin(adata, keys = ['n_genes', 'total_counts'] , groupby = 'Sample', log = True, stripplot = False, 
             multi_panel = True, rotation = 90, save = figure + '_violin_plots_samples.pdf')
plt.savefig('./Figure_plots/'+ figure + '/' + figure + '_violin_plots_samples.pdf', bbox_inches='tight')
```

## dendrogram

```python
with plt.rc_context({'figure.figsize': (20, 5)}):
    sc.pl.dendrogram(adata, groupby = clusteringlayer, save = figure + '_dendrogram.pdf')
    plt.savefig('./Figure_plots/'+ figure + '/' + figure + '_dendrogram.pdf')
```
