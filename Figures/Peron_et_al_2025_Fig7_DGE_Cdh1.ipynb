### Comparisson between Cdh1 and H2B

```python
def plot_lfc_comparison(df1, df2,
                        lfc_col='log2FoldChange',
                        padj_col='padj',
                        padj_thresh=0.1,
                        label1='Cdh1',
                        label2='H2B',
                        ax=None):  
    # prepare tables
    d1 = df1[[lfc_col, padj_col]].rename(
        columns={lfc_col: 'lfc1', padj_col: 'padj1'}
    )
    d2 = df2[[lfc_col, padj_col]].rename(
        columns={lfc_col: 'lfc2', padj_col: 'padj2'}
    )
    df = d1.join(d2, how='inner')
    # determine significance categories
    sig1 = df['padj1'] < padj_thresh
    sig2 = df['padj2'] < padj_thresh
    df['category'] = 'none'
    df.loc[sig1 & ~sig2, 'category'] = 'cdh1_only'
    df.loc[~sig1 & sig2, 'category'] = 'h2b_only'
    df.loc[sig1 & sig2, 'category'] = 'both'
    colors = {
        'none': 'lightgrey',
        'cdh1_only': 'dodgerblue',
        'h2b_only': 'darkviolet',
        'both': 'red'
    }
    if ax is None:
        fig, ax = plt.subplots(figsize=(6,6))

    for cat, color in colors.items():
        subset = df[df['category'] == cat]
        ax.scatter(
            subset['lfc1'],
            subset['lfc2'],
            color=color,
            s=20,
            alpha=0.7,
            label=cat.replace('_', ' '), rasterized=True
        )

    ax.axhline(0, color='black', linewidth=1.5)
    ax.axvline(0, color='black', linewidth=1.5)

    for spine in ax.spines.values():
        spine.set_linewidth(1.5)

    lims = [
        min(df['lfc1'].min(), df['lfc2'].min()),
        max(df['lfc1'].max(), df['lfc2'].max())
    ]
    ax.plot(lims, lims, 'k--', alpha=0.4)

    # correlation
    r, _ = stats.spearmanr(df['lfc1'], df['lfc2'])

    ax.tick_params(axis='both', labelsize=20)

    # Remove grid
    ax.grid(False)

    # Remove legend
    # ax.legend(frameon=False)  ← removed

    # Add r value as text annotation instead
    ax.text(
        0.05, 0.95,
        f"Spearman: {r:.2f}",
        transform=ax.transAxes,
        fontsize=22,
        va='top', ha='left'
    )

    if ax is None:
        plt.tight_layout()
        plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 6), sharey=True)

plot_lfc_comparison(
    s_neo_c, s_neo_h,
    padj_thresh=0.05,
    label1="Cdh1 RNAi neoblasts",
    label2="H2B RNA neoblasts",
    ax=axes[0]
)

plot_lfc_comparison(
    s_prog_c, s_prog_h,
    padj_thresh=0.05,
    label1="Cdh1 RNAi progenitors",
    label2="H2B RNA progenitors",
    ax=axes[1]
)

plot_lfc_comparison(
    s_diff_c, s_diff_h,
    padj_thresh=0.05,
    label1="Cdh1 RNAi differentiated",
    label2="H2B RNA differentiated",
    ax=axes[2]
)

plt.tight_layout()
plt.savefig('./Figure_plots/' + figure + "/correlation_score.pdf",
            format="pdf", bbox_inches="tight")
plt.show()
```
